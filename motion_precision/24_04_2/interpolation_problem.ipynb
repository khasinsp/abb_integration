```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd
from urdfpy import URDF
from urdfpy import matrix_to_rpy
import plotly.express as px
```

```python
com_df = pd.read_csv("com.csv")
com_ts = com_df["timestamp"].values
com_ts = (com_ts - com_ts[0]) / 1000.0
com_pos = com_df.drop(columns=["timestamp"]).values

# 1) DataFrame anlegen
df = pd.DataFrame({
    'timestamp': com_ts,
    'com_pos_1': com_pos[:, 0],
    'com_pos_2': com_pos[:, 1],
    'com_pos_3': com_pos[:, 2],
    'com_pos_4': com_pos[:, 3],
    'com_pos_5': com_pos[:, 4],
    'com_pos_6': com_pos[:, 5],

})

# 1) In Long-Form umwandeln
df_long = df.melt(
    id_vars='timestamp',
    value_vars=df.columns[1:],
    var_name='Channel',
    value_name='command axis [deg]'
)

# 2) Plot bauen
fig = px.line(
    df_long,
    x='timestamp',
    y='command axis [deg]',
    color='Channel',
    title="Achse über Zeit",
    labels={'timestamp': 'Timestamp'}
)

# 3) Achsen anpassen
fig.update_xaxes(
    tickformat="%H:%M:%S\n%d-%m-%Y",
)

fig.show()
```
