```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
act_df = pd.read_csv("act.csv")
com_df = pd.read_csv("com.csv")

act_df = act_df[act_df["timestamp"].isin(com_df["timestamp"])]
com_df = com_df[com_df["timestamp"].isin(act_df["timestamp"])]

act_j = act_df.drop(columns=["timestamp"]).values
com_j = com_df.drop(columns=["timestamp"]).values
j_err = act_j - com_j
```

```python
print(j_err.max())
```
